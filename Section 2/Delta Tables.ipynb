```python
%sql
USE CATALOG hive_metastore
```

```python
%sql
CREATE TABLE employees
    (id INT, name STRING, salary DOUBLE)
```

```python
%sql
INSERT INTO employees
VALUES
    (1, "Adam", 3500.00),
    (2, "Sarah", 4020.5);

INSERT INTO employees
VALUES
    (3, "John", 2999.3),
    (4, "Thomas", 4000.3);

INSERT INTO employees
VALUES
    (5, "Anna", 2500.0);

INSERT INTO employees
VALUES
    (6, "Kim", 6200.3)
```

```python
%sql
DESCRIBE DETAIL employees
```

```python
%sql
UPDATE employees
SET salary = salary + 100
WHERE name LIKE "A%"
```
